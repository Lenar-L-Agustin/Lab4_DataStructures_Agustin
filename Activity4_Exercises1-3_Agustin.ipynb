```python
LAST_NAME = "Agustin"
STUDENT_ID = "TUPM-26-1937"
FAVORITE_ARTIST = "Maroon 5"

SEED_NUM = int(STUDENT_ID[-1])
codes = [ord(ch) % 10 for ch in FAVORITE_ARTIST]

# Check Value
print("Inputs")
print(f"LAST_NAME: {LAST_NAME}")
print(f"SEED_NUM: {SEED_NUM}")
print(f"FAVORITE_ARTIST: {FAVORITE_ARTIST}")
print(f"Character codes: {codes}")
```

```python
# EXERCISE 1
print("-" * 10, "EXERCISE 1", "-" * 10)
# Generate Test Data
sources = [LAST_NAME + "_T" + str(n) for n in range(1, 6)]
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

records = []
for i in range(5):
    records.append((sources[i], categories[i], SEED_NUM * (i + 1) + codes[i % len(codes)]))
records.append(records[0])  # repeated record
for k in range(5, 9):  # extra measurements (TEMPERATURE and VOLTAGE)
    records.append((sources[k % 5], categories[k % 2], SEED_NUM * 2 + codes[k % len(codes)] * 3 + k))

print("1. Generated Test Data")
for i in range(0, len(records), 2):
    print(records[i:i + 2])

# Organize by source then category
organized = {}
for s, c, v in records:
    if s not in organized:
        organized[s] = {}
    if c not in organized[s]:
        organized[s][c] = []
    organized[s][c].append(v)

print("2. Organized Test Data")
for s in organized:
    print(s, organized[s])

# Repeated and distinct
pairs = [r[:2] for r in records]
repeated = set([p for p in pairs if pairs.count(p) > 1])
values = [r[2] for r in records]

print("3. Repeated / Distinct Data")
print("Repeated:", repeated)
print("Distinct pairs:", len(set(pairs)), "| Distinct values:", sorted(set(values)))

# Summaries per category
summary = []
print("4. Numerical and Categorical Summaries")
for c in categories:
    vals = [r[2] for r in records if r[1] == c]
    avg = round(sum(vals) / len(vals), 2)
    print(f"{c}: count={len(vals)} total={sum(vals)} avg={avg} min={min(vals)} max={max(vals)}")
    summary.append((avg, c))

print("5. Processed Results (avg, category)")
print(summary)

summary.sort()
summary.reverse()
print("6. Ordered Results (highest average first)")
print([s[1] for s in summary])

final = {"operator": LAST_NAME, "records": len(records), "repeated": len(repeated),
         "total": sum(values), "average": round(sum(values) / len(values), 2),
         "max": max(values), "min": min(values), "top_category": summary[0][1]}
print("7. Final Summary")
print(final)
```

```python
# EXERCISE 2
print("-" * 10, "EXERCISE 2", "-" * 10)
# Generate Performance Records
students = [LAST_NAME + "_S" + str(n) for n in range(1, 6)]
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]

records = []
for k in range(10):
    score = 50 + (SEED_NUM * 7 + codes[k % len(codes)] * 5 + k * 3) % 51
    records.append((students[k // 2], subjects[(k // 2 + k % 2) % 5], score))
records.append(records[0])  # repeated record
records.append((students[2], subjects[4], None))  # incomplete record
print("1. Generated Performance Records")
for i in range(0, len(records), 2):
    print(records[i:i + 2])

# Organize by student then subject (valid scores only)
valid = [r for r in records if r[2] is not None]
organized = {}
for st, sj, sc in valid:
    if st not in organized:
        organized[st] = {}
    if sj not in organized[st]:
        organized[st][sj] = []
    organized[st][sj].append(sc)

print("2. Organized Student/Subject Data")
for st in organized:
    print(st, organized[st])

# Repeated and incomplete
pairs = [r[:2] for r in records]
repeated = set([p for p in pairs if pairs.count(p) > 1])
incomplete = [r[:2] for r in records if r[2] is None]

print("3. Repeated / Incomplete Records")
print("Repeated:", repeated)
print("Incomplete:", incomplete)

# Performance categories
graded = []
for st, sj, sc in valid:
    if sc >= 90: label = "EXCELLENT"
    elif sc >= 80: label = "VERY GOOD"
    elif sc >= 70: label = "GOOD"
    elif sc >= 60: label = "FAIR"
    else: label = "NEEDS IMPROVEMENT"
    graded.append((sc, label))
labels = [g[1] for g in graded]

print("4. Performance Categories (score, label)")
print(graded)
print([(n, labels.count(n)) for n in ["EXCELLENT", "VERY GOOD", "GOOD", "FAIR", "NEEDS IMPROVEMENT"]])

# Summaries
student_sum = []
for st in students:
    vals = [r[2] for r in valid if r[0] == st]
    student_sum.append((round(sum(vals) / len(vals), 2), st, len(vals)))
subject_sum = []
for sj in subjects:
    vals = [r[2] for r in valid if r[1] == sj]
    subject_sum.append((round(sum(vals) / len(vals), 2), sj, len(vals)))

print("5. Student and Subject Summaries (avg, name, records)")
print(student_sum)
print(subject_sum)

student_sum.sort()
student_sum.reverse()
subject_sum.sort()
subject_sum.reverse()
print("6. Ordered Results (highest average first)")
print([s[1] for s in student_sum])
print([s[1] for s in subject_sum])

scores = [r[2] for r in valid]
report = {"operator": LAST_NAME, "records": len(records), "valid": len(scores),
          "incomplete": len(incomplete), "repeated": len(repeated),
          "class_avg": round(sum(scores) / len(scores), 2), "high": max(scores),
          "low": min(scores), "top_student": student_sum[0][1], "top_subject": subject_sum[0][1]}
print("Final Student Performance Report")
print(report)
```

```python
# EXERCISE 3
print("-" * 10, "EXERCISE 3", "-" * 10)
# Generate Transaction Records
customers = [LAST_NAME + "_C" + str(n) for n in range(1, 6)]
products = ["NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR"]
prices = {"NOTEBOOK": 45, "PEN": 12, "USB": 350, "HEADPHONES": 890, "CALCULATOR": 650}

trans = []
for t in range(8):
    qty = 1 + (SEED_NUM + codes[t % len(codes)]) % 5
    trans.append((customers[t % 5], products[(t * 2 + SEED_NUM) % 5], qty))
trans.append(trans[0]) # repeated purchase

print("1. Generated Transaction Records")
for i in range(0, len(trans), 2):
    print(trans[i:i + 2])

# Organize by customer then product
organized = {}
for c, p, q in trans:
    if c not in organized:
        organized[c] = {}
    if p not in organized[c]:
        organized[c][p] = []
    organized[c][p].append(q)

print("2. Organized Transaction Data")
for c in organized:
    print(c, organized[c])

# Repeated and distinct
pairs = [t[:2] for t in trans]
repeated = set([p for p in pairs if pairs.count(p) > 1])

print("3. Repeated / Distinct Transactions")
print("Repeated:", repeated)
print("Distinct customers:", len(set([t[0] for t in trans])), "| Distinct products:", len(set([t[1] for t in trans])))

# Transaction values
values = [t[2] * prices[t[1]] for t in trans]
print("4. Transaction Values (same order as above)")
print(values, "| Overall:", sum(values))

# Summaries
cust_sum = []
for cu in customers:
    cust_sum.append((sum([t[2] * prices[t[1]] for t in trans if t[0] == cu]), cu, sum([t[2] for t in trans if t[0] == cu])))
prod_sum = []
for pr in products:
    prod_sum.append((sum([t[2] * prices[pr] for t in trans if t[1] == pr]), pr, sum([t[2] for t in trans if t[1] == pr])))

print("5. Customer Summaries (spent, name, items)")
print(cust_sum)
print("6. Product Summaries (revenue, name, units)")
print(prod_sum)

cust_sum.sort()
cust_sum.reverse()
prod_sum.sort()
prod_sum.reverse()
ordered_values = values.copy()
ordered_values.sort()
ordered_values.reverse()
print("7. Ordered Results (highest first)")
print([s[1] for s in cust_sum])
print([s[1] for s in prod_sum])
print(ordered_values)

report = {"operator": LAST_NAME, "transactions": len(trans), "items": sum([t[2] for t in trans]),
          "grand_total": sum(values), "average": round(sum(values) / len(values), 2),
          "largest": ordered_values[0], "smallest": ordered_values[-1],
          "repeated": len(repeated), "top_customer": cust_sum[0][1], "top_product": prod_sum[0][1]}
print("8. Overall Transaction Summary")
print(report)
```
